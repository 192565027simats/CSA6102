```python
import re

# --------------------------------------------------
# Experiment 1.3 - Ransomware IoCs and Signatures
# --------------------------------------------------

print("=" * 60)
print("RANSOMWARE IoCs AND SIGNATURE ANALYSIS")
print("=" * 60)

# --------------------------------------------------
# Sample Ransomware Database
# --------------------------------------------------

ransomware_db = [
    {
        "family": "WannaCry",
        "domains": [
            "iuqerfsodp9ifjaposdfjhgosurijfaewrwergwea.com"
        ],
        "ips": [
            "192.168.1.25"
        ],
        "extensions": [
            ".wncry",
            ".wcry"
        ],
        "note": """
Oops! Your files have been encrypted.
Send 300 USD in Bitcoin.
You have 3 days to pay.
Decrypt your files now.
"""
    },

    {
        "family": "LockBit",
        "domains": [
            "lockbit.support"
        ],
        "ips": [
            "45.67.12.90"
        ],
        "extensions": [
            ".lockbit"
        ],
        "note": """
All your files are encrypted.
Contact us immediately.
Bitcoin payment required.
Recover your files.
"""
    }
]

# --------------------------------------------------
# Parse IoCs
# --------------------------------------------------

total_domains = 0
total_ips = 0
total_extensions = 0

print("\nRANSOMWARE FAMILY PROFILES\n")

for family in ransomware_db:

    print("-" * 60)
    print("Family :", family["family"])

    print("\nDomains:")
    for domain in family["domains"]:
        print(" •", domain)
        total_domains += 1

    print("\nIP Addresses:")
    for ip in family["ips"]:
        print(" •", ip)
        total_ips += 1

    print("\nEncrypted File Extensions:")
    for ext in family["extensions"]:
        print(" •", ext)
        total_extensions += 1

    print("\nRansom Note:")
    print(family["note"].strip())

    print("\nCharacteristics:")

    note = family["note"].lower()

    if "bitcoin" in note:
        print(" ✓ Bitcoin payment requested")

    if "decrypt" in note or "recover" in note:
        print(" ✓ File recovery message found")

    if "encrypted" in note:
        print(" ✓ File encryption warning")

    if re.search(r"\d+\s*day", note):
        print(" ✓ Payment deadline detected")

# --------------------------------------------------
# IoC Collection
# --------------------------------------------------

print("\n" + "=" * 60)
print("IOC COLLECTION")
print("=" * 60)

print("Total Families        :", len(ransomware_db))
print("Total Domains         :", total_domains)
print("Total IP Addresses    :", total_ips)
print("Total File Extensions :", total_extensions)

# --------------------------------------------------
# YARA-Like Rules
# --------------------------------------------------

print("\n" + "=" * 60)
print("YARA-LIKE DETECTION RULES")
print("=" * 60)

for family in ransomware_db:

    print(f"\nrule {family['family']}_Detection")
    print("{")
    print("    strings:")

    for i, ext in enumerate(family["extensions"], start=1):
        print(f'        $ext{i} = "{ext}"')

    print('        $btc = "Bitcoin"')
    print('        $enc = "encrypted"')

    print("\n    condition:")
    print("        any of them")
    print("}")

# --------------------------------------------------
# Simulated Ransomware Deployment Timeline
# --------------------------------------------------

print("\n" + "=" * 60)
print("SIMULATED RANSOMWARE DEPLOYMENT TIMELINE")
print("=" * 60)

timeline = [
    ("09:00", "Initial Access"),
    ("09:02", "Malware Downloaded"),
    ("09:04", "Persistence Established"),
    ("09:05", "Privilege Escalation"),
    ("09:06", "Command & Control Communication"),
    ("09:08", "File Encryption Started"),
    ("09:10", "Ransom Note Created"),
    ("09:12", "Encryption Completed")
]

for time, event in timeline:
    print(f"{time}  -->  {event}")

print("\n" + "=" * 60)
print("ANALYSIS COMPLETE")
print("=" * 60)
```
