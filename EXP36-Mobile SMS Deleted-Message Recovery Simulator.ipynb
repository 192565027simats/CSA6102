```python
# Deleted SMS Recovery Simulation

def recover_deleted_messages(sms_table):
    """
    Returns deleted messages that are still recoverable.
    A message is recoverable if it is deleted but not overwritten.
    """

    recoverable = [
        row
        for row in sms_table
        if row["is_deleted"] and not row["overwritten"]
    ]

    return recoverable


def summarize_table(sms_table):
    """
    Provides a summary of active, recoverable,
    and permanently lost messages.
    """

    active = [
        row
        for row in sms_table
        if not row["is_deleted"]
    ]

    recoverable = recover_deleted_messages(sms_table)

    lost = [
        row
        for row in sms_table
        if row["is_deleted"] and row["overwritten"]
    ]

    return {
        "active": len(active),
        "recoverable": len(recoverable),
        "permanently_lost": len(lost)
    }


# ---------------- Sample Input ----------------

sms_table = [
    {
        "rowid": 1,
        "address": "+1-555-0101",
        "body": "See you at 6pm",
        "date": "2026-01-01",
        "is_deleted": False,
        "overwritten": False
    },
    {
        "rowid": 2,
        "address": "+1-555-0199",
        "body": "Transfer the funds now, delete after reading",
        "date": "2026-01-02",
        "is_deleted": True,
        "overwritten": False
    },
    {
        "rowid": 3,
        "address": "+1-555-0150",
        "body": "Old spam message",
        "date": "2025-06-01",
        "is_deleted": True,
        "overwritten": True
    },
]


# ---------------- Run Recovery ----------------

recoverable = recover_deleted_messages(sms_table)


# ---------------- Verification ----------------

assert len(recoverable) == 1
assert recoverable[0]["rowid"] == 2
assert "Transfer the funds" in recoverable[0]["body"]

summary = summarize_table(sms_table)

assert summary == {
    "active": 1,
    "recoverable": 1,
    "permanently_lost": 1
}


print("All test cases passed.\n")

print("Recoverable Deleted Messages:")

for message in recoverable:
    print("\nRow ID :", message["rowid"])
    print("From   :", message["address"])
    print("Message:", message["body"])

print("\nSMS Table Summary:")
print(summary)
```
