```python
import os

def make_fake_jpeg(payload: bytes):
    return b"\xff\xd8\xff" + payload + b"\xff\xd9"

# Create two fake JPEG files
jpeg1 = make_fake_jpeg(b"PHOTO_OF_SUSPECT_CAR")
jpeg2 = make_fake_jpeg(b"CCTV_FRAME_CAPTURE")

# Simulate a raw disk containing random data and embedded JPEGs
raw_disk_blob = (
    os.urandom(30) +
    jpeg1 +
    os.urandom(40) +
    jpeg2 +
    os.urandom(20)
)

def carve_jpegs(blob: bytes):
    recovered = []
    start_marker = b"\xff\xd8\xff"
    end_marker = b"\xff\xd9"

    pos = 0

    while True:
        start = blob.find(start_marker, pos)
        if start == -1:
            break

        end = blob.find(end_marker, start)
        if end == -1:
            break

        end += len(end_marker)
        recovered.append(blob[start:end])

        pos = end

    return recovered

# Recover JPEG files
recovered_files = carve_jpegs(raw_disk_blob)

print("Recovered", len(recovered_files), "file(s) via carving")

# Display recovered file contents
for i, data in enumerate(recovered_files, start=1):
    print(f"Recovered File {i}: {data}")
```
