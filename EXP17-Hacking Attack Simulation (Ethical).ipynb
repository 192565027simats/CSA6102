```python
import time
import re

print("=" * 65)
print("      EXPERIMENT 1.4 - HACKING ATTACK SIMULATION (ETHICAL)")
print("=" * 65)

# ============================================================
# PART 1 - Password Brute Force Simulation (Rate Limited)
# ============================================================

print("\n1. PASSWORD BRUTE FORCE SIMULATION")
print("-" * 50)

actual_password = "admin123"

password_list = [
    "123456",
    "password",
    "admin",
    "welcome",
    "admin123",
    "letmein"
]

attempts = 0
attack_logs = []

for pwd in password_list:
    attempts += 1
    print(f"Attempt {attempts}: Trying '{pwd}'")
    attack_logs.append(f"Attempt {attempts}: {pwd}")

    if pwd == actual_password:
        print("Password Matched (Simulation)")
        attack_logs.append("Password Matched")
        break

    time.sleep(0.5)      # Ethical rate limiting

# ============================================================
# PART 2 - Weak Password Analysis
# ============================================================

print("\n2. PASSWORD STRENGTH ANALYSIS")
print("-" * 50)

passwords = [
    "123456",
    "password",
    "qwerty",
    "Admin@123",
    "Cyber@2026",
    "abc123"
]

for pwd in passwords:

    score = 0

    if len(pwd) >= 8:
        score += 1

    if re.search(r"[A-Z]", pwd):
        score += 1

    if re.search(r"[a-z]", pwd):
        score += 1

    if re.search(r"\d", pwd):
        score += 1

    if re.search(r"[!@#$%^&*]", pwd):
        score += 1

    if score <= 2:
        strength = "Weak"
    elif score == 3:
        strength = "Medium"
    else:
        strength = "Strong"

    print(f"{pwd:15s} --> {strength}")

# ============================================================
# PART 3 - SQL Injection Detection
# ============================================================

print("\n3. SQL INJECTION PAYLOAD DETECTION")
print("-" * 50)

payloads = [
    "' OR '1'='1",
    "admin' --",
    "SELECT * FROM users",
    "normalUser",
    "'; DROP TABLE users; --"
]

sql_patterns = [
    r"or\s+'.*'='.*'",
    r"--",
    r"drop\s+table",
    r"union\s+select",
    r"select\s+\*",
]

for payload in payloads:

    detected = False

    for pattern in sql_patterns:
        if re.search(pattern, payload, re.IGNORECASE):
            detected = True
            break

    if detected:
        print(f"Detected SQL Injection : {payload}")
        attack_logs.append(f"SQL Injection -> {payload}")
    else:
        print(f"Safe Input             : {payload}")

# ============================================================
# PART 4 - Directory Traversal Detection
# ============================================================

print("\n4. DIRECTORY TRAVERSAL DETECTION")
print("-" * 50)

paths = [
    "../etc/passwd",
    "../../windows/system32",
    "images/logo.png",
    "..\\..\\secret.txt",
    "documents/report.pdf"
]

for path in paths:

    if "../" in path or "..\\" in path:
        print(f"Traversal Detected : {path}")
        attack_logs.append(f"Traversal -> {path}")
    else:
        print(f"Safe Path          : {path}")

# ============================================================
# PART 5 - Attack Pattern Logs
# ============================================================

print("\n" + "=" * 65)
print("ATTACK PATTERN LOGS")
print("=" * 65)

for log in attack_logs:
    print(log)

# ============================================================
# PART 6 - Vulnerability Classification Report
# ============================================================

print("\n" + "=" * 65)
print("VULNERABILITY CLASSIFICATION REPORT")
print("=" * 65)

print("Brute Force Attack        : HIGH")
print("Weak Passwords            : HIGH")
print("SQL Injection             : CRITICAL")
print("Directory Traversal       : HIGH")

# ============================================================
# PART 7 - WAF Rule Suggestions
# ============================================================

print("\n" + "=" * 65)
print("WAF RULE SUGGESTIONS")
print("=" * 65)

rules = [
    "Block requests containing '../' or '..\\'",
    "Block SQL keywords: SELECT, UNION, DROP, OR",
    "Limit login attempts to 5 per minute",
    "Enforce strong password policy",
    "Log repeated failed login attempts",
    "Block requests containing '--' comments"
]

for i, rule in enumerate(rules, start=1):
    print(f"Rule {i}: {rule}")

print("\n" + "=" * 65)
print("SIMULATION COMPLETED SUCCESSFULLY")
print("=" * 65)
```
